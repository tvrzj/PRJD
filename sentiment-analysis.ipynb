```python
import os
import json
import torch
import random
import numpy as np
import plotly.express as px
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tokenizer = AutoTokenizer.from_pretrained("siebert/sentiment-roberta-large-english")
model = AutoModelForSequenceClassification.from_pretrained("siebert/sentiment-roberta-large-english")

model.to(device)
model.eval()

seed = 42  # You can choose any integer as a seed
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(seed)
```

## own sentences test 

```python
text = ['I like you very much.',
 'I hate you very much',
 'I love you',
 'You fucking smell',
 'You have a beautiful hair.',
 'I eat dead babies.',
 'I find ancient philosophy unapealing.',
 'I would like to kill you',
 'There is a cat on the mat.',
 'I feel neutral',
 'I am slightly irritated',
 'This is sad',
 'A sad clown made me chuckle',
 'I am not unhappy'
 ]

tens_data = []

for sent in text:
    inputs = tokenizer(sent, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

    with torch.no_grad():
        logits = model(**inputs).logits
        #print(sent, logits.argmax().item(), logits.tolist())

        entry = {
        'sentence': sent,
        'color': 'red' if logits.argmax().item() == 0 else 'green',
        'x': logits.tolist()[0][0],
        'y': logits.tolist()[0][1],
        'line': np.sum(logits.tolist()),
        'label': logits.argmax().item()
        }
        tens_data.append(entry)

        probabilities = torch.nn.functional.softmax(logits, dim=1)
        predicted_class = torch.argmax(probabilities, dim=1)

        print(probabilities)
        print(predicted_class)

movie_df = pd.DataFrame(tens_data)
movie_df.sort_values(by='line')
```

## MetAI test

```python
model.to(device)
model.eval()

directory = '/srv/data/MetAI/ai-sentences'

# Initialize lists to store data
keys = []
colors = []
x_values = []
y_values = []
mean_values = []
sentences = []
line_counter = []

# Iterate over all files in the specified directory
for filename in os.listdir(directory):

    # Check if the file is a .txt file
    if filename.endswith('.txt'):
        file_path = os.path.join(directory, filename)
        
        # Open the .txt file and read its lines
        with open(file_path, 'r') as file:

            line_counter = 0

            for line in file: 
                inputs = tokenizer(line, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

                with torch.no_grad():  # Disable gradient calculation for inference
                    logits = model(**inputs).logits                
                    argmax = logits.argmax().item()
                    
                keys.append(filename.rstrip(".txt"))
                colors.append('red' if argmax == 0 else 'green')
                x_values.append(logits.tolist()[0][0])
                y_values.append(logits.tolist()[0][1])
                mean_values.append(np.mean(logits.tolist()))
                sentences.append(line)
                sent_number.append(line_counter)
                line_counter += 1

AI_sentiment_df = pd.DataFrame({
    'archive_key': keys,
    'color': colors,
    'x': x_values,
    'y': y_values,
    '2d': mean_values,
    'sentence': sentences
})
```

```python
# create sum of tensors and append it as 6th column
AI_sentiment_df['2d-sum'] = AI_sentiment_df['x'] + AI_sentiment_df['y']
AI_sentiment_df.insert(5, '2d-sum', AI_sentiment_df.pop('2d-sum'))
```

```python
AI_sentiment_df['label'] = AI_sentiment_df['color'].apply(lambda x: 'positive' if x == 'green' else 'negative')
```

```python
# map year data from metadata_df
metadata_df= pd.read_csv('../data/metadata_df.csv')

# Step 1: Create a mapping from archive_key to year
year_mapping = dict(zip(metadata_df['archive_key'], metadata_df['year']))

# Step 2: Create a new column in AI_sentiment_df for the year
AI_sentiment_df['year'] = AI_sentiment_df['archive_key'].map(year_mapping)
```
